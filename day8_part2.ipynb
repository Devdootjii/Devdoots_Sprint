2 The function in a program lets a play a game and returns the Score as an
integer. You need to read a file 'Hi-score.txt' which is either blank or contains the
previous Hi-score. You need to write a program to update the Hi-score whenever the
game() function breaks the Hi-score.

```python
def update_hi_score(current_score):
    # 1. File se purana score padhna
    try:
        with open("Hi-score.txt", "r") as f:
            content = f.read().strip()
            # Agar file khali hai, toh hi-score 0 maan lo
            hi_score = int(content) if content else 0
    except FileNotFoundError:
        # Agar file exist hi nahi karti, toh hi-score 0 hai
        hi_score = 0

    # 2. Compare karna aur update karna
    if current_score > hi_score:
        print(f"Naya High Score! Purana tha: {hi_score}, Ab hai: {current_score}")
        with open("Hi-score.txt", "w") as f:
            f.write(str(current_score))
    else:
        print(f"High score nahi toota. High score abhi bhi {hi_score} hai.")

# Example usage:
# score = game()  <-- Tumhara game function yahan score return karega
score = 150 
update_hi_score(score)
```

3. Write a program to generate multiplication tables from 2 to 20 and write it to the
different files. Place these files in a folder for a 13 — year old.
A file contains a word "Donkey" multiple times.

```python
import os

# 1. Folder ka naam rakhna
folder_name = "Student_Tables"

# 2. Folder banana (agar nahi bana hai toh)
if not os.path.exists(folder_name):
    os.makedirs(folder_name)

# 3. 2 se 20 tak table banane ka loop
for i in range(2, 21):
    # Har table ke liye nayi file ka naam: Table_2.txt, Table_3.txt...
    filename = f"{folder_name}/Table_{i}.txt"
    
    with open(filename, "w") as f:
        for j in range(1, 11):
            f.write(f"{i} x {j} = {i * j}\n")

print("Saari tables folder mein save ho gayi hain!")
```


4.
You need to write a program which replace this word With by updating the file.

```python
# File ka naam jismein se word replace karna hai
filename = "notes.txt"

# 1. Pehle file ko read mode ("r") mein open karo
with open(filename, "r") as f:
    data = f.read()

# 2. String replace function ka use karo
# Syntax: old_word.replace("kise_badalna_hai", "kiske_saath_badalna_hai")
updated_data = data.replace("Donkey", "######")

# 3. Ab file ko write mode ("w") mein open karke update kar do
with open(filename, "w") as f:
    f.write(updated_data)

print("File successfully update ho gayi hai!")
```

5. Repeat program 4 for a list of such words to be censored.

```python
# 1. File aur wo words ki list jise censor karna hai
filename = "notes.txt"
censor_list = ["Donkey", "Monkey", "Gadha", "Fool"]

# 2. File ko read karo
with open(filename, "r") as f:
    content = f.read()

# 3. Har word ke liye loop chalao aur replace karo
for word in censor_list:
    content = content.replace(word, "######")

# 4. Update ki hui content ko file mein wapas likho
with open(filename, "w") as f:
    f.write(content)

print("Saare words successfully censor ho gaye hain!")
```
