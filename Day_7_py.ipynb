Question 4: Write a program to find whether a given number is prime or not.

```python
num=int(input("Enter a number:"))

factorial=1

if num<0:
    print("Sorry,factorial are not exist for negative number")

elif num==0:
    print("the factorial of 0 is 1")

else:
    for i in range(1,num+1):
        factorial=factorial*i

print(f"the factorial of {num} is {factorial}")
```

Question 5: Write a program to find the sum of first n number using while loop


```python
n=int(input("Enter number:"))

sum=0
i=1

while i<=n:
    sum=sum+1
    i += 1
    print(f"the sum of first {n} natural number is :{sum}")
```
