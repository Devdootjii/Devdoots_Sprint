Question: Ek 2-D vector ki class banao aur uska use karke dusri class banao jo 3-D vector ko represent kare.

```python
class Vector2D:
    def __init__(self, i, j):
        self.i = i
        self.j = j

    def show(self):
        print(f"{self.i}i + {self.j}j")

# Vector3D inherit kar raha hai Vector2D ko
class Vector3D(Vector2D):
    def __init__(self, i, j, k):
        super().__init__(i, j) # super() purani class se i aur j ko load karega
        self.k = k

    def show(self):
        print(f"{self.i}i + {self.j}j + {self.k}k")

# Testing
v2 = Vector2D(2, 5)
v2.show()  # Output: 2i + 5j

v3 = Vector3D(1, 4, 9)
v3.show()  # Output: 1i + 4j + 9k
```

Question: Ek class 'Pets' banao 'Animals' class se, aur fir ek class 'Dog' banao 'Pets' class se. Dog class mein ek 'bark' method jodo.

```python
class Animals:
    def __init__(self):
        print("Animal created")

class Pets(Animals):
    def __init__(self):
        super().__init__()
        print("Pet created")

class Dog(Pets):
    def __init__(self):
        super().__init__()
        print("Dog created")
    
    def bark(self):
        print("Woof! Woof! 🐕")

# Testing
d = Dog()
d.bark()  # Output: Woof! Woof!
```

Question: Ek class 'Employee' banao jisme salary aur increment properties hon. Ek @property decorator ke sath 'salaryAfterIncrement' method banao, aur ek 'setter' method bhi likho jo salary ke hisab se increment value ko change kare.

```python
class Employee:
    def __init__(self, salary, increment):
        self.salary = salary
        self.increment = increment  # Ye flat bonus ya percentage ho sakta hai

    @property
    def salaryAfterIncrement(self):
        return self.salary + self.increment

    @salaryAfterIncrement.setter
    def salaryAfterIncrement(self, new_total_salary):
        # Agar final salary badli, toh naya increment calculate hoga
        self.increment = new_total_salary - self.salary

# Testing
e = Employee(50000, 5000)
print("Pehle total salary thi:", e.salaryAfterIncrement) # Output: 55000

# Ab humne direct variable ki tarah value change ki
e.salaryAfterIncrement = 62000 
print("Naya increment automatic calculate hua:", e.increment) # Output: 12000
```

Question: Complex numbers ($a + ib$) ko represent karne ke liye ek 'Complex' class banao. + aur * operators ko overload karo jo unhe add aur multiply kar sakein.

```python
class Complex:
    def __init__(self, r, i):
        self.real = r
        self.imaginary = i

    # '+' operator overload kiya
    def __add__(self, num):
        return Complex(self.real + num.real, self.imaginary + num.imaginary)

    # '*' operator overload kiya
    def __mul__(self, num):
        real_part = (self.real * num.real) - (self.imaginary * num.imaginary)
        img_part = (self.real * num.imaginary) + (self.imaginary * num.real)
        return Complex(real_part, img_part)

    def show(self):
        print(f"{self.real} + {self.imaginary}i")

# Testing
c1 = Complex(3, 2)
c2 = Complex(1, 7)

c3 = c1 + c2  # Automatic __add__ chalega
c3.show()     # Output: 4 + 9i
```

Question 5, 6, 7 (Combined): Ek n-dimensions ka Vector class banao. + aur * (dot product) overload karo. __str__ jodo jisse vector $7i + 8j + 10k$ format mein print ho (assume dimension 3). Aur __len__ method override karo jo vector ki dimension return kare

```python
class Vector:
    def __init__(self, vec_list):
        self.vec = vec_list  

    def __add__(self, other):
        new_vec = []
        for i in range(len(self.vec)):
            new_vec.append(self.vec[i] + other.vec[i])
        return Vector(new_vec)

    def __mul__(self, other):
        dot_product = 0
        for i in range(len(self.vec)):
            dot_product += self.vec[i] * other.vec[i]
        return dot_product

    def __str__(self):
        return f"{self.vec}i + {self.vec}j + {self.vec}k"

    def __len__(self):
        return len(self.vec)

# ---- TESTING AREA (Yahan check karo) ----
v1 = Vector()  # Ab error nahi aayegi, data pass ho gaya hai!
v2 = Vector()

print("Vector 1:", v1)          
print("Length of v1:", len(v1)) 

v3 = v1 + v2
print("Sum of Vectors:", v3)    

print("Dot Product:", v1 * v2)
```
